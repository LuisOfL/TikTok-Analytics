# Exportar datos a sql


```python
import pandas as pd
```

```python
df = pd.read_csv("clean_data.csv")
df.head()
```

```python
df["description"] = df["description"].fillna("Sin descripción")
df.isnull().sum()
```

### Cambiar nombre de variables conforme el modelo dimencional

```python
df = df.rename(columns={
    "author": "author_name",
    "create_time": "fecha_completa",
})
```

```python
df.head()
```

## Crear dimenciones de mi modelo estrella

#### Dimencion author

```python
Dim_Author = df[["author_name"]].reset_index(drop=True)
```

```python
Dim_Author.head()
```

#### Dimencion video

```python
Dim_Video = df[[
    "video_url",
    "description",
    "hashtags",
    "music"
]].reset_index(drop=True)
```

```python
Dim_Video = Dim_Video.applymap(lambda x: str(x).encode("utf-8", "ignore").decode("utf-8"))
```

```python
Dim_Video.head()
```

### Dimencion fecha

```python
df["fecha_completa"] = pd.to_datetime(df["fecha_completa"])

Dim_Tiempo = df[["fecha_completa"]].copy()
Dim_Tiempo["anio"] = Dim_Tiempo["fecha_completa"].dt.year
Dim_Tiempo["trimestre"] = Dim_Tiempo["fecha_completa"].dt.quarter
Dim_Tiempo["mes"] = Dim_Tiempo["fecha_completa"].dt.month
Dim_Tiempo["nombre_mes"] = Dim_Tiempo["fecha_completa"].dt.strftime("%B")
Dim_Tiempo["fecha_completa"] = Dim_Tiempo["fecha_completa"].dt.date

Dim_Tiempo = Dim_Tiempo.reset_index(drop=True)
```

```python
Dim_Tiempo.head()
```

### Tabla de hechos

```python
Fact_TikTok_Metrics = df[["plays","shares","comments","likes"]].reset_index(drop=True)
Fact_TikTok_Metrics["id_video"] = Fact_TikTok_Metrics.index + 1
Fact_TikTok_Metrics["id_author"] = Fact_TikTok_Metrics.index + 1
Fact_TikTok_Metrics["id_date"] = Fact_TikTok_Metrics.index + 1
Fact_TikTok_Metrics = Fact_TikTok_Metrics[["id_video","id_author","id_date","plays","shares","comments","likes"]]
```

```python
Fact_TikTok_Metrics.head()
```

## Exportar a SQL

```python
Dim_Author.to_csv("Dim_Author.csv", index=False, encoding="utf-8")
Dim_Video.to_csv(
    "Dim_Video.csv",
    index=False,
    encoding="utf-8-sig"
)
Dim_Tiempo.to_csv("Dim_Tiempo.csv", index=False, encoding="utf-8")
Fact_TikTok_Metrics.to_csv("Fact_TikTok_Metrics.csv", index=False, encoding="utf-8")
```
